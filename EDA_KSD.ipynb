```python
# Identify possible outliers
outlier_columns = ["price",
    "bedrooms",
    "bathrooms",
    "sqft_living",
    "sqft_lot",
    "floors",
    "sqft_above",
    "sqft_basement",
    "yr_built",
    "grade",
    "condition"]

plt.figure(figsize=(16, 8))

for i, column in enumerate(outlier_columns, 1):
    plt.subplot(3, 4, i)
    sns.boxplot(x=df_clean[column])
    plt.title(f"Boxplot: {column}")

plt.tight_layout()
plt.show()
```

```python
# IQR method to count potential outliers

for column in outlier_columns:
    q1 = df_clean[column].quantile(0.25)
    q3 = df_clean[column].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_count = df_client[
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    ].shape[0]

    print(f"{column}: {outlier_count} potential outliers")
```

```python
# Define the 10% most expensive houses
top_10_threshold = df_clean["price"].quantile(0.90)
df_clean["top_10_percent"] = df_clean["price"] >= top_10_threshold
```

```python
# Correlation analysis, top 10 features correlated with price

correlation_matrix = df_clean.select_dtypes(include="number").corr()

price_correlations = (
    correlation_matrix["price"]
    .sort_values(ascending=False)
    .to_frame(name="correlation_with_price")
    .head(10)
)
top_features = price_correlations.index.tolist()
display(price_correlations)
plt.figure(figsize=(10, 8))

sns.heatmap(
    df_clean[top_features].corr(),
    cmap="coolwarm",
    center=0,
    annot=True,
    fmt=".2f",
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title("Top Features Correlated with Price")
plt.show()
```

```python
#H1: Central houses are more likely to be in the top 10% of prices
# Number and percentage of top-10%-priced houses in Central vs. Outskirts
top_10 = (
    df_clean[df_clean["top_10_percent"]]
    .groupby("area_type")
    .size()
    .reset_index(name="number_of_top_10_houses")
)

# Percentage of all top-10% houses located in each area
top_10["percentage_of_all_top_10_houses"] = (
    top_10["number_of_top_10_houses"]
    / top_10["number_of_top_10_houses"].sum()
    * 100
).round(2)

# Percentage within each geographic area
area_totals = df_clean.groupby("area_type").size()

top_10["percentage_within_area"] = (
    top_10.apply(
        lambda row: row["number_of_top_10_houses"] / area_totals[row["area_type"]] * 100,
        axis=1
    )
).round(2)

display(top_10)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=top_10,
    x="area_type",
    y="percentage_within_area",
    hue="area_type",
    legend=False
)

plt.title("Share of Top 10% Priced Houses by Area")
plt.xlabel("Area type")
plt.ylabel("Top 10% houses within area (%)")
plt.show()
```

```python
# Geographic distribution of Top 10% highest-priced houses
# Calculate the dataset geographic centre
central_lat = df_clean["lat"].median()
central_long = df_clean["long"].median()

# Approximate distance to the dataset centre in kilometres
df_clean["distance_to_centre_km"] = (
    np.sqrt(
        (df_clean["lat"] - central_lat) ** 2 +
        (df_clean["long"] - central_long) ** 2
    ) * 111
)

# Data-driven definition: closest 25% of properties are "Central"
central_cutoff = df_clean["distance_to_centre_km"].quantile(0.25)

df_clean["area_type"] = np.where(
    df_clean["distance_to_centre_km"] <= central_cutoff,
    "Central",
    "Outskirts"
)

print(f"Dataset geographic centre: ({central_lat:.4f}, {central_long:.4f})")
print(f"Central-area cutoff: {central_cutoff:.2f} km")
```

```python
#H2: High-value central houses are sold across several periods

import matplotlib.dates as mdates
from matplotlib.ticker import MaxNLocator

# Monthly sales counts
high_value_central = df_clean[
    (df_clean["area_type"] == "Central") &
    (df_clean["top_10_percent"])
]

sales_by_month = (
    high_value_central
    .groupby(pd.Grouper(key="date", freq="MS"))
    .size()
    .rename("number_of_sales")
    .reset_index()
)

# Include every month in the date range
all_months = pd.date_range(
    start=df_clean["date"].min().to_period("M").to_timestamp(),
    end=df_clean["date"].max().to_period("M").to_timestamp(),
    freq="MS"
)

sales_by_month = (
    sales_by_month
    .set_index("date")
    .reindex(all_months, fill_value=0)
    .rename_axis("period")
    .reset_index()
)

display(sales_by_month)

# Improved time-series chart
fig, ax = plt.subplots(figsize=(14, 6))

sns.lineplot(
    data=sales_by_month,
    x="period",
    y="number_of_sales",
    marker="o",
    markersize=8,
    linewidth=2.5,
    color="#EE1313",
    ax=ax
)

# Add a subtle fill below the trend line
ax.fill_between(
    sales_by_month["period"],
    sales_by_month["number_of_sales"],
    alpha=0.15,
    color="#C67676"
)

# Ensure the y-axis shows whole numbers only
max_sales = sales_by_month["number_of_sales"].max()
ax.set_ylim(0, max_sales * 1.15 + 1)
ax.yaxis.set_major_locator(MaxNLocator(integer=True))

# Clear month labels
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=1))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b\n%Y"))

ax.set_title(
    "Monthly Sales of High-Value Houses in Central Areas",
    fontsize=16,
    fontweight="bold",
    pad=15
)
ax.set_xlabel("Sale month", fontsize=12)
ax.set_ylabel("Number of sales", fontsize=12)

ax.grid(axis="y", linestyle="--", alpha=0.5)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
#H3: Outskirts houses are more likely to have average prices

#median_price = df_clean["price"].median()
# print(f"Median house price: ${median_price:,.2f}")
# Define "average-priced" as within ±25% of the median
#lower_price = median_price - (median_price * 0.25)
# upper_price = median_price + (median_price * 0.25)
lower_price = df_clean["price"].quantile(0.25)
upper_price = df_clean["price"].quantile(0.75)

df_clean["average_price"] = df_clean["price"].between(
    lower_price,
    upper_price
)

h3_results = (
    df_clean.groupby("area_type")["average_price"]
    .mean()
    .mul(100)
    .reset_index(name="average_price")
)

display(h3_results)

sns.barplot(data=h3_results, x="area_type", y="average_price")
plt.title("Average-Priced Houses by Area")
plt.ylabel("Percentage of houses (%)")
plt.show()
```
